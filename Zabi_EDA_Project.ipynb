```python
def simplify_devtype(x):
    if pd.isna(x):
        return "Unknown"

    x = x.lower()

    if "developer" in x:
        return "Developer"
    elif any(word in x for word in ["data", "scientist", "ai/ml"]):
        return "Data/AI"
    elif any(word in x for word in ["devops", "cloud", "system administrator"]):
        return "Infrastructure"
    elif any(word in x for word in ["architect", "engineering manager"]):
        return "Architecture/Management"
    elif "cybersecurity" in x or "infosec" in x:
        return "Security"
    elif "student" in x:
        return "Student"
    elif "researcher" in x:
        return "Research/Education"
    elif any(word in x for word in ["manager", "executive", "founder", "product manager"]):
        return "Management/Business"
    else:
        return "Other"

df["DevType_Grouped"] = df["DevType"].apply(simplify_devtype)
```

```python
def group_industry(x):
    if pd.isna(x):
        return "Unknown"
    
    if x == "Software Development":
        return "Technology"
    elif x in ["Internet, Telecomm or Information Services",
               "Computer Systems Design and Services"]:
        return "Technology"
    elif x in ["Fintech", "Banking/Financial Services", "Insurance"]:
        return "Finance"
    elif x in ["Healthcare"]:
        return "Healthcare"
    elif x in ["Higher Education", "Government"]:
        return "Public/Education"
    elif x in ["Retail and Consumer Services",
               "Transportation, or Supply Chain",
               "Manufacturing",
               "Energy",
               "Media & Advertising Services"]:
        return "Other Industries"
    else:
        return "Other"

ai_df["Industry_Grouped"] = ai_df["Industry"].apply(group_industry)
```

```python
def group_complex(x):
    if pd.isna(x):
        return "Unknown"
    
    if "Very well" in x or "Good" in x:
        return "Good"
    elif "Bad" in x or "poor" in x:
        return "Poor"
    elif "Neither" in x:
        return "Neutral"
    elif "don't use" in x or "don't know" in x:
        return "No experience"
    else:
        return "Other"

ai_df["AIComplex_Grouped"] = ai_df["AIComplex"].apply(group_complex)
```

```python
def simplify_frustration(x):
    if pd.isna(x):
        return "No response"
    if "almost right" in x:
        return "AI output issues"
    elif "Debugging" in x:
        return "Debugging difficulty"
    elif "less confident" in x:
        return "Reduced confidence"
    elif "hard to understand" in x:
        return "Understanding difficulty"
    elif "I haven’t encountered" in x:
        return "No problems"
    elif "I don’t use" in x:
        return "Doesn't use AI"
    else:
        return "Other"

ai_df["AIFrustration_Grouped"] = ai_df["AIFrustration"].apply(simplify_frustration)
```

```python
def simplify_model(x):
    if pd.isna(x):
        return "Unknown"
    
    models = []
    
    if "openAI" in x:
        models.append("OpenAI")
    if "Gemini" in x:
        models.append("Gemini")
    if "Anthropic" in x:
        models.append("Anthropic")
    if "DeepSeek" in x:
        models.append("DeepSeek")
    if "Meta Llama" in x:
        models.append("Meta Llama")
    if "Mistral" in x:
        models.append("Mistral")
    if "Microsoft Phi" in x:
        models.append("Microsoft")
    if "Perplexity" in x:
        models.append("Perplexity")
    if "Amazon" in x:
        models.append("Amazon")
    if "Alibaba" in x:
        models.append("Alibaba")
    if "Cohere" in x:
        models.append("Cohere")
    if "X Grok" in x:
        models.append("Grok")
        
    return ";".join(models) if models else "Other"

ai_df["AIModelsHaveWorkedWith"] = ai_df["AIModelsHaveWorkedWith"].apply(simplify_model)
```

## Does experience relate to AI adoption?

```python
experience_ai = pd.crosstab(
    ai_df["WorkExp"],
    ai_df["AISelect"],
    normalize="index"
) * 100

experience_ai.plot(
    kind="line",
    figsize=(10,6),
    marker="o",
    title="AI Adoption Across Years of Work Experience"
)

plt.xlabel("Years of Work Experience")
plt.ylabel("Percentage")
plt.show()
```

```python
bins = [-1, 2, 5, 10, 20, 100]
labels = ["0–2", "3–5", "6–10", "11–20", "20+"]

ai_df["WorkExp_Group"] = pd.cut(
    ai_df["WorkExp"],
    bins=bins,
    labels=labels
)

experience_ai = pd.crosstab(
    ai_df["WorkExp_Group"],
    ai_df["AISelect"],
    normalize="index"
) * 100

experience_ai.plot(
    kind="bar",
    stacked=True,
    figsize=(10,6),
    title="AI Adoption by Work Experience"
)

plt.xlabel("Years of Experience")
plt.ylabel("Percentage")
plt.xticks(rotation=0)
plt.show()
```
